```python
#Exercise 1 – Add.py

n1 = float(input("first number  ")) #get first input number and convert to float
n2 = float(input("second number  ")) #get second input number and convert to float

def add_fn():
  return "the sum is " + str(n1 + n2) #add numbers and return it

add_fn() #call it
```

```python
#Exercise 2 – CountDown.py

n = int(input("give me a natural number")) #get input number and convert to int

while n > 0: #ensure that n is a natural number
  print(str(n)) #write out the current number
  n = n - 1 #subtract n by 1
```

```python
#Exercise 3 – SortList.py

user_input = input("give me a list of comma seperated words ") #get list of input words

words = [word.strip() for word in user_input.split(",")] #convert input into list

words.sort() #sort words

print(words) #print output
```

```python
#Exercise 4 – Distance.py

from math import sqrt #get sqrt func from math lib

coord1_input = input("give me one 2D coordinate ") #get first coords
coord2_input = input("give me one 2D coordinate ") #get second coords

def input_to_coords(coord):
  return [float(number.strip()) for number in coord.split(",")] #split input into coords and convert into floats

coord1 = input_to_coords(coord1_input) #assign value
coord2 = input_to_coords(coord2_input) #assign value

distance = sqrt((coord1[0] - coord2[0])**2 + (coord1[1] - coord2[1])**2) #use distance formula

print(f'distance is {distance}') #print output
```
