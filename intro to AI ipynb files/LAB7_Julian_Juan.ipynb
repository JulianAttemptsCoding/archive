## Task 1: Text Preprocessing Using spaCy



```python
# ========================= TASK 1A: ORIGINAL INPUT ========================= # This header marks the first spaCy preprocessing example.
import spacy  # Import spaCy so we can load the language model for preprocessing.
nlp = spacy.load("en_core_web_sm")  # Load the English spaCy model into memory.

original_text = "Natural Language Processing (NLP) is a fascinating field of AI that enables computers to understand human language."  # Store the original sentence in a variable.
original_doc = nlp(original_text)  # Process the sentence with spaCy so it can be tokenized and lemmatized.

original_tokens = [token.text for token in original_doc]  # Save every token exactly as spaCy split it.
original_clean_tokens = [token.text.lower() for token in original_doc if not token.is_punct and not token.is_space]  # Save lowercase tokens while removing punctuation and blank spaces.
original_lemmas = [token.lemma_.lower() for token in original_doc if not token.is_punct and not token.is_space]  # Save lowercase lemma forms while removing punctuation and blank spaces.

print("Original sentence:")  # Print a label before showing the original sentence.
print(original_text)  # Print the original sentence itself.
print()  # Print a blank line to make the output easier to read.
print("All tokens:")  # Print a label before showing every token.
print(original_tokens)  # Print the full token list.
print()  # Print a blank line for spacing.
print("Clean tokens (lowercase, no punctuation):")  # Print a label before showing the cleaned token list.
print(original_clean_tokens)  # Print the cleaned token list.
print()  # Print a blank line for spacing.
print("Lemmas (base word forms):")  # Print a label before showing the lemma list.
print(original_lemmas)  # Print the lemma list.
print()  # Print a blank line for spacing.
print("Token details:")  # Print a label before showing a line-by-line token breakdown.

for token in original_doc:  # Loop through every token in the processed spaCy document.
    print(f"Text: {token.text:<15} | Lemma: {token.lemma_:<15} | Stop Word: {token.is_stop}")  # Print the token text, lemma, and whether spaCy marks it as a stop word.
```

**What this block does:**  
This block sends the original sentence into spaCy, splits it into tokens, removes punctuation for a cleaner list, and converts each word to its lemma. The last loop prints a detailed breakdown so you can see the original word, its base form, and whether it is a stop word.

**What to notice in the output:**  
- Tokenization breaks the sentence into smaller pieces such as words and punctuation.
- Lemmatization changes words to their base form, such as **enables → enable**.
- The detailed table helps you see exactly what spaCy is doing to each token.

## Task 2: Sentiment Analysis Using a Fine-Tuned Model

This task uses a pretrained sentiment model. It predicts whether a sentence is **Positive**, **Neutral**, or **Negative** and also gives a confidence score.

```python
# ========================= TASK 2A: ORIGINAL INPUTS ========================= # This header marks the original sentiment-analysis examples.
from transformers import pipeline  # Import pipeline so we can load a ready-made sentiment model.

sentiment_model = pipeline("sentiment-analysis", model="cardiffnlp/twitter-roberta-base-sentiment")  # Load the fine-tuned sentiment model from Hugging Face.
label_map = {"LABEL_0": "Negative", "LABEL_1": "Neutral", "LABEL_2": "Positive"}  # Convert the model's label IDs into readable sentiment names.

text_samples = [  # Create a list of the original example sentences.
    "I love this product! It's amazing.",  # First sentence that should look positive.
    "The service was terrible and I am very disappointed.",  # Second sentence that should look negative.
    "The movie was okay, not the best but not the worst.",  # Third sentence that may be neutral or mixed.
]  # End the list of original sentiment examples.

sentiment_counts = {"Positive": 0, "Neutral": 0, "Negative": 0}  # Create a dictionary to count how many results fall into each sentiment category.

for text in text_samples:  # Loop through each example sentence one at a time.
    prediction = sentiment_model(text)  # Ask the model to predict the sentiment of the current sentence.
    predicted_label = label_map[prediction[0]["label"]]  # Convert the raw label into a readable word like Positive or Negative.
    confidence_score = round(prediction[0]["score"], 4)  # Round the confidence score to four decimal places so it is easier to read.
    sentiment_counts[predicted_label] += 1  # Increase the count for the predicted sentiment category by one.
    print("Text:", text)  # Print the sentence that was analyzed.
    print("Predicted sentiment:", predicted_label)  # Print the predicted sentiment label.
    print("Confidence score:", confidence_score)  # Print the model's confidence score.
    print("-" * 60)  # Print a divider line so each result is easy to separate visually.

print("Sentiment distribution summary:", sentiment_counts)  # Print the final count of positive, neutral, and negative predictions.
```

**What this block does:**  
This block loads a fine-tuned sentiment model and tests it on three sample sentences. For each sentence, it prints the predicted sentiment label and the model's confidence score.

**Important note for your explanation:**  
This model predicts **sentiment**, not detailed emotions like joy, fear, or anger. In this lab, “different emotions” really means different sentiment categories such as positive, neutral, and negative.

```python
# ========================= TASK 2B: MODIFIED INPUT ========================= # This header marks the modified sentiment-analysis example.
from transformers import pipeline  # Import pipeline again so this cell can run on its own if needed.

sentiment_model = pipeline("sentiment-analysis", model="cardiffnlp/twitter-roberta-base-sentiment")  # Load the same fine-tuned sentiment model.
label_map = {"LABEL_0": "Negative", "LABEL_1": "Neutral", "LABEL_2": "Positive"}  # Create the same label-to-name mapping.

modified_text = "The food was delicious but overpriced."  # Store the teacher-suggested modified sentence in a variable.
modified_prediction = sentiment_model(modified_text)  # Run the sentiment model on the modified sentence.
modified_label = label_map[modified_prediction[0]["label"]]  # Convert the raw model label into a readable sentiment name.
modified_confidence = round(modified_prediction[0]["score"], 4)  # Round the model's confidence score to four decimal places.

print("Modified text:", modified_text)  # Print the modified sentence.
print("Predicted sentiment:", modified_label)  # Print the predicted sentiment label for the modified sentence.
print("Confidence score:", modified_confidence)  # Print the confidence score for the modified sentence.
print()  # Print a blank line for readability.
print("Why this example is interesting:")  # Print a label before the explanation section.
print("This sentence contains both a positive idea ('delicious') and a negative idea ('overpriced').")  # Explain why the sentence may be harder for the model to classify.
print("Because the sentence is mixed, the model may choose Neutral, Positive, or Negative depending on which words it weighs more strongly.")  # Explain why the output may change compared with simpler sentences.
```

**What this block does:**  
This block tests the sentiment model on a mixed sentence that contains both positive and negative language.

**How to explain the result:**  
- Confidence scores often change when a sentence is more mixed or ambiguous.  
- A clearly positive or clearly negative sentence usually gets a stronger confidence score.  
- If the output seems odd, you can say the sentence contains mixed sentiment, which can make classification harder.

## Task 3: Named Entity Recognition (NER)

NER finds important names in text, such as **people**, **organizations**, and **locations**.

```python
# ========================= TASK 3A: ORIGINAL INPUT ========================= # This header marks the original named-entity-recognition example.
import warnings  # Import warnings so we can hide non-critical warning messages.
from transformers import pipeline  # Import pipeline so we can load a ready-made NER model.

warnings.filterwarnings("ignore")  # Hide extra warnings so the output is cleaner for screenshots.
ner_pipeline = pipeline("ner", model="dbmdz/bert-large-cased-finetuned-conll03-english", aggregation_strategy="simple")  # Load the NER model and merge subword pieces into whole entities.

text_sample = "Apple Inc. was founded by Steve Jobs and Steve Wozniak in California. The company made $274.5 billion in revenue in 2020."  # Store the original NER example sentence.
entities = ner_pipeline(text_sample)  # Run named entity recognition on the original sentence.

print("Original text:")  # Print a label before showing the original text.
print(text_sample)  # Print the original text.
print()  # Print a blank line for spacing.
print("Detected entities:")  # Print a label before showing the entity results.

for entity in entities:  # Loop through each detected entity in the result list.
    entity_word = entity["word"]  # Save the entity text, such as Apple Inc. or California.
    entity_type = entity["entity_group"]  # Save the entity category, such as ORG, PER, or LOC.
    entity_score = round(entity["score"], 4)  # Round the confidence score so it is easier to read.
    print(f"Entity: {entity_word:<20} | Type: {entity_type:<5} | Confidence: {entity_score}")  # Print the entity, its type, and its confidence score.
```

**What this block does:**  
This block uses an NER model to locate named entities in the sentence. The model returns the detected entity text, the entity type, and a confidence score.

**Common entity types you may see:**  
- **PER** = person  
- **ORG** = organization  
- **LOC** = location

```python
# ========================= TASK 3B: MODIFIED INPUT ========================= # This header marks the modified named-entity-recognition example.
import warnings  # Import warnings again so this cell can run independently.
from transformers import pipeline  # Import pipeline again so this cell can run independently.

warnings.filterwarnings("ignore")  # Hide warning messages again for cleaner output.
ner_pipeline = pipeline("ner", model="dbmdz/bert-large-cased-finetuned-conll03-english", aggregation_strategy="simple")  # Load the same NER model again.

modified_text = "Elon Musk founded SpaceX in California."  # Store the teacher-suggested modified NER sentence.
modified_entities = ner_pipeline(modified_text)  # Run named entity recognition on the modified sentence.

print("Modified text:")  # Print a label before showing the modified text.
print(modified_text)  # Print the modified sentence.
print()  # Print a blank line for spacing.
print("Detected entities:")  # Print a label before showing the new entity results.

for entity in modified_entities:  # Loop through each entity detected in the modified sentence.
    entity_word = entity["word"]  # Save the entity text.
    entity_type = entity["entity_group"]  # Save the entity category.
    entity_score = round(entity["score"], 4)  # Round the confidence score for readability.
    print(f"Entity: {entity_word:<20} | Type: {entity_type:<5} | Confidence: {entity_score}")  # Print the entity result in a readable format.

print()  # Print a blank line before the comparison notes.
print("What changed?")  # Print a label before the comparison notes.
print("Different names lead to different detected entities.")  # Explain that changing the sentence changes the detected named entities.
print("Confidence scores can rise or fall depending on how certain the model is about each entity.")  # Explain how confidence scores relate to model certainty.
```

**What this block does:**  
This block repeats NER with a different sentence so you can compare how entity detection changes when the person, company, and location names change.

**How to explain the result:**  
- The model should detect **Elon Musk** as a person, **SpaceX** as an organization, and **California** as a location.  
- Confidence scores show how sure the model is. A higher score usually means the model is more certain.

## Task 4: Machine Translation Using Generative AI

This task translates English into Spanish.

```python
# ============================ TASK 4A: ORIGINAL INPUT ============================  # This header marks the original machine-translation example.

from transformers import MarianMTModel  # Import MarianMTModel so we can load a translation model directly.
from transformers import MarianTokenizer  # Import MarianTokenizer so we can tokenize English text for the Marian model.

model_name = "Helsinki-NLP/opus-mt-en-es"  # Store the English-to-Spanish Marian model name in a variable.
tokenizer = MarianTokenizer.from_pretrained(model_name)  # Load the tokenizer that matches the Marian translation model.
model = MarianMTModel.from_pretrained(model_name)  # Load the Marian English-to-Spanish translation model.

original_sentence = "Natural language processing is a branch of AI."  # Store the original English sentence in a variable.
original_text_list = [original_sentence]  # Put the sentence into a list because batch translation methods expect a list of texts.

tokenized_inputs = tokenizer(original_text_list, return_tensors="pt", padding=True)  # Convert the English text into PyTorch tensors for the model.
generated_tokens = model.generate(**tokenized_inputs)  # Ask the model to generate the translated Spanish output tokens.
decoded_outputs = tokenizer.batch_decode(generated_tokens, skip_special_tokens=True)  # Convert the generated token IDs back into readable text.
original_translation = decoded_outputs[0]  # Take the first translated sentence from the decoded output list.

print("Original English sentence:")  # Print a label before showing the English sentence.
print(original_sentence)  # Print the original English sentence.
print()  # Print a blank line for readability.
print("Spanish translation:")  # Print a label before showing the translated sentence.
print(original_translation)  # Print the translated Spanish sentence.
```

**What this block does:**  
This block loads a machine-translation model and translates the original English sentence into Spanish.

**What to look for:**  
- Check whether the meaning stays the same.  
- Check whether the Spanish grammar looks natural.  
- Minor wording differences are normal because translation models can choose different valid phrases.

## Task 5: Topic Modeling and Classification Using AI

This task uses **zero-shot classification**, which means the model tries to choose the best category label even though it was not trained only on your exact class examples.

```python
# ========================= TASK 5A: ORIGINAL INPUTS ========================= # This header marks the original zero-shot topic-classification examples.
from transformers import pipeline  # Import pipeline so we can load a zero-shot classification model.

topic_model_pipeline = pipeline("zero-shot-classification", model="facebook/bart-large-mnli")  # Load the zero-shot classification model.
articles = {  # Create a dictionary of article names and short article texts.
    "Article 1": "The new government policy on climate change was announced today.",  # Politics-related example.
    "Article 2": "The stock market has seen an unexpected rise this week.",  # Finance-related example.
    "Article 3": "Scientists discovered a new exoplanet that could support life.",  # Science-related example.
    "Article 4": "The latest smartphone has innovative features never seen before.",  # Technology-related example.
}  # End the article dictionary.

categories = ["Politics", "Finance", "Science", "Technology"]  # Create the list of candidate topic labels the model can choose from.

print("Original article classifications:")  # Print a label before showing the classification results.
print("-" * 80)  # Print a divider line for readability.

for article_name, article_text in articles.items():  # Loop through each article name and article text in the dictionary.
    classification = topic_model_pipeline(article_text, candidate_labels=categories)  # Ask the model to score the article against all candidate topic labels.
    best_label = classification["labels"][0]  # Save the top predicted topic label.
    best_score = round(classification["scores"][0], 4)  # Save the top confidence score and round it for readability.
    print("Article name:", article_name)  # Print the article name.
    print("Article text:", article_text)  # Print the article text.
    print("Predicted category:", best_label)  # Print the best predicted category.
    print("Confidence score:", best_score)  # Print the confidence score for the best category.
    print("-" * 80)  # Print a divider line after each article result.
```

**What this block does:**  
This block classifies each short article into one of four categories: Politics, Finance, Science, or Technology.

**What to notice in the output:**  
- The model chooses the category with the highest score.  
- Higher confidence usually means the article matches that topic more clearly.  
- Very short or ambiguous articles may get lower confidence scores.

```python
# ========================= TASK 5B: MODIFIED INPUTS ========================= # This header marks the modified zero-shot topic-classification examples.
from transformers import pipeline  # Import pipeline again so this cell can run independently.

topic_model_pipeline = pipeline("zero-shot-classification", model="facebook/bart-large-mnli")  # Load the same zero-shot classification model again.
categories = ["Politics", "Finance", "Science", "Technology"]  # Recreate the candidate topic labels.

modified_article = "Bitcoin prices reached an all-time high this week."  # Store the teacher-suggested modified article in a variable.
modified_result = topic_model_pipeline(modified_article, candidate_labels=categories)  # Classify the modified article against the candidate topic labels.

mixed_topic_article = "The government is debating cryptocurrency taxes while tech companies expand their AI systems."  # Create an extra mixed-topic article to show what happens when one text contains more than one topic.
mixed_topic_result = topic_model_pipeline(mixed_topic_article, candidate_labels=categories)  # Classify the mixed-topic article against the same candidate topic labels.

print("Modified article:")  # Print a label before showing the teacher-suggested modified article.
print(modified_article)  # Print the modified article text.
print("Top predicted category:", modified_result["labels"][0])  # Print the highest-scoring category for the modified article.
print("Top confidence score:", round(modified_result["scores"][0], 4))  # Print the highest confidence score for the modified article.
print()  # Print a blank line for spacing.
print("All category scores for the modified article:")  # Print a label before showing all category-score pairs.
for label, score in zip(modified_result["labels"], modified_result["scores"]):  # Loop through every label and score pair returned by the classifier.
    print(f"{label}: {round(score, 4)}")  # Print each label with its score so the ranking is easy to see.
print()  # Print a blank line for spacing.
print("Mixed-topic article:")  # Print a label before showing the mixed-topic article.
print(mixed_topic_article)  # Print the mixed-topic article text.
print("Top predicted category:", mixed_topic_result["labels"][0])  # Print the highest-scoring category for the mixed-topic article.
print("Top confidence score:", round(mixed_topic_result["scores"][0], 4))  # Print the highest confidence score for the mixed-topic article.
print()  # Print a blank line for spacing.
print("All category scores for the mixed-topic article:")  # Print a label before showing all scores for the mixed-topic example.
for label, score in zip(mixed_topic_result["labels"], mixed_topic_result["scores"]):  # Loop through every label and score pair for the mixed-topic example.
    print(f"{label}: {round(score, 4)}")  # Print each label with its score so you can see how mixed-topic text affects the ranking.
```

**What this block does:**  
This block tests topic classification on a Bitcoin article and also on a mixed-topic article.

**How to explain the result:**  
- The Bitcoin sentence should usually be classified as **Finance**.  
- Confidence scores change depending on how strongly the text matches one topic.  
- If an article contains multiple topics, the model still chooses one top label, but the other category scores may also be relatively high.
